```python
import os
import sys
```

## Setup
We need to find the Raven python handler and bring that into our path. Possibly this could be bypassed by adding raven to the pythonpath.

```python
frameworkDir = os.path.abspath(os.path.expanduser('~/projects/raven'))
```

```python
if frameworkDir not in sys.path:
    sys.path.append(frameworkDir)
```

This next step loads (lazily, sometimes) the libraries for Raven into memory, in addition to making the PythonRaven object available to this interpreter.

```python
from framework import Raven
```

Now we can instantiate an instance of the Python handler for RAVEN.

```python
raven = Raven()
```

## Load
This instance can't do much without a workflow to use, so let's load up a workshop example. This will include reading the XML and initializing the underlying Simulation object.

```python
targetWorkflow = os.path.join(frameworkDir, 'doc', 'workshop', 'forwardSampling', 'Inputs', 'test_1_sample_and_plot.xml')
raven.loadWorkflowFromXML(targetWorkflow)
```

## Run
Now we can run the workflow. It takes 1000 MonteCarlo samples of the Projectile model, so it takes a second or two. Fun note: the MessageHandler has been counting time ever since it was initialized!

```python
raven.runWorkflow()
```

## Analyze
Now we can look at some of the data produced and, for example, plot it. Let's get the DataObject named "results".

```python
results = raven.getEntity('DataObjects', 'results').asDataset()
results.plot.scatter(x="v0", y="angle", hue="r")
%matplotlib inline
import matplotlib.pyplot as plt
# plt.show()
```
